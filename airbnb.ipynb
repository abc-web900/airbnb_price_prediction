```python
# ================================================================
# AIRBNB PRICE PREDICTION — ONE COLAB CELL, STARTING FROM RAW CSV
# Preserves the final notebook's 12 predictors and engineered features.
# Changes: fold-fitted preprocessing, optional host-grouped splitting,
# fair training-only trimming, raw/log comparisons, complete holdout.
# All reported price metrics are in ORIGINAL price units, not log units.
# ================================================================

# --------------------------- 1. SETTINGS --------------------------
DATA_PATH = "/content/listings.csv"
OUTPUT_DIR = "/content/airbnb_results"
TARGET = "price"
SEED = 2026

# True: evaluate listings from hosts absent from training.
# False: random listing split, like the original notebook.
GROUP_BY_HOST = True
HOLDOUT_SIZE = 0.20  # Fraction of HOSTS if grouped; otherwise of rows.
CV_FOLDS = 3
SEARCH_ITERATIONS = 8  # Per boosted model; increase for a broader search.
BOOSTING_ROUNDS = 350
RF_TREES = 200
N_GEO_CLUSTERS = 20
MAX_CATEGORIES = 128  # Per categorical column; bounds dense one-hot memory.
RUN_VOTING = True
RUN_STACKING = True  # Additional inner-CV fits; set False to reduce work.
RUN_PERMUTATION = True
PERMUTATION_ROWS = 2000
PERMUTATION_REPEATS = 3
PRIMARY_METRIC = "MAE"  # "MAE" or "RMSE"; lower is better.

# Use a FRESH Colab runtime: do not run earlier imputation cells first.
import sys, os, gc, json, time, subprocess, warnings
import importlib.metadata as metadata
from pathlib import Path
from packaging.version import Version

minimum_versions = {
    "numpy": "1.24", "pandas": "2.0", "scikit-learn": "1.6",
    "lightgbm": "4.6.0", "xgboost": "3.1.3", "catboost": "1.2.8",
    "cloudpickle": "3.1", "matplotlib": "3.7",
}
needed = []
for package, minimum in minimum_versions.items():
    try:
        installed = metadata.version(package)
    except metadata.PackageNotFoundError:
        installed = "0"
    if Version(installed) < Version(minimum):
        needed.append(f"{package}>={minimum}")
if needed:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", *needed])

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cloudpickle
from IPython.display import display
from sklearn.base import BaseEstimator, TransformerMixin, RegressorMixin, clone
from sklearn.utils.validation import check_is_fitted
from sklearn.model_selection import (
    train_test_split, GroupShuffleSplit, GroupKFold, KFold,
    cross_validate, RandomizedSearchCV,
)
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer, make_column_selector
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.cluster import KMeans
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor, VotingRegressor, StackingRegressor
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, mean_squared_log_error,
    median_absolute_error, r2_score,
)
from sklearn.inspection import permutation_importance
from lightgbm import LGBMRegressor
from xgboost import XGBRegressor
from catboost import CatBoostRegressor
from threadpoolctl import threadpool_limits

# Suppress only this benign LightGBM/sklearn naming warning.
# Training, convergence and numerical warnings remain visible.
warnings.filterwarnings("ignore", message=(
    "X does not have valid feature names, but LGBMRegressor was fitted with feature names"
))

N_JOBS = max(1, min(4, os.cpu_count() or 1))
# Also bound BLAS/OpenMP threads for linear models and geographic clustering.
_thread_limit = threadpool_limits(limits=N_JOBS)
OUT = Path(OUTPUT_DIR)
OUT.mkdir(parents=True, exist_ok=True)
if PRIMARY_METRIC not in {"MAE", "RMSE"}:
    raise ValueError("PRIMARY_METRIC must be 'MAE' or 'RMSE'.")
if CV_FOLDS < 2 or SEARCH_ITERATIONS < 1:
    raise ValueError("Use at least two CV folds and one search iteration.")

NUM_COLS = [
    "latitude", "longitude", "accommodates", "bathrooms", "bedrooms",
    "beds", "minimum_nights", "availability_365", "number_of_reviews",
]
CAT_COLS = ["neighbourhood_cleansed", "property_type", "room_type"]
RAW_FEATURES = NUM_COLS + CAT_COLS
GROUP_COL = "__host_group"  # Splitting only; NEVER a model predictor.

# -------------------- 2. FEATURE ENGINEERING ----------------------
class AirbnbFeatureEngineer(TransformerMixin, BaseEstimator):
    def __init__(self, engineered=True, geo=True, missing_flags=True,
                 n_clusters=20, random_state=2026):
        self.engineered = engineered
        self.geo = geo
        self.missing_flags = missing_flags
        self.n_clusters = n_clusters
        self.random_state = random_state

    def _clean(self, X):
        if not isinstance(X, pd.DataFrame):
            raise TypeError("Pass raw inputs as a pandas DataFrame.")
        missing = sorted(set(RAW_FEATURES) - set(X.columns))
        if missing:
            raise ValueError(f"Missing predictor columns: {missing}")
        z = X[RAW_FEATURES].copy()  # Drops IDs and grouping metadata.
        for c in NUM_COLS:
            z[c] = pd.to_numeric(z[c], errors="coerce").astype(float)
        z[NUM_COLS] = z[NUM_COLS].replace([np.inf, -np.inf], np.nan)
        z["latitude"] = z["latitude"].where(z["latitude"].between(-90, 90))
        z["longitude"] = z["longitude"].where(z["longitude"].between(-180, 180))
        for c in CAT_COLS:
            z[c] = z[c].astype("string").str.strip().replace("", pd.NA)
        return z

    def fit(self, X, y=None):
        z = self._clean(X)
        # These statistics are learned from the CURRENT training fold only.
        self.medians_ = z[NUM_COLS].median().fillna(0.0)
        self.n_features_in_ = X.shape[1]
        if self.geo:
            geo = z[["latitude", "longitude"]].fillna(self.medians_).to_numpy()
            k = min(self.n_clusters, len(np.unique(geo, axis=0)))
            if k < 1:
                raise ValueError("No rows available for geographic clustering.")
            self.kmeans_ = KMeans(
                n_clusters=k, n_init=10, random_state=self.random_state
            ).fit(geo)
        return self

    def transform(self, X):
        check_is_fitted(self, "medians_")
        z = self._clean(X)
        # Capture missingness BEFORE median filling.
        if self.missing_flags:
            flags = z[RAW_FEATURES].isna()
            for c in RAW_FEATURES:
                z[f"{c}__missing"] = flags[c].astype(np.float32)
            z["missing_count"] = flags.sum(axis=1).astype(np.float32)

        z[NUM_COLS] = z[NUM_COLS].fillna(self.medians_)
        for c in CAT_COLS:
            z[c] = z[c].fillna("Unknown").astype(object)

        if self.engineered:
            denominator = z["accommodates"].where(z["accommodates"] > 0)
            for source in ["beds", "bedrooms", "bathrooms"]:
                z[f"{source}_per_guest"] = z[source] / denominator
            # Missing counts use fold medians, not an assumption of zero space.
            z["capacity_space"] = z[["beds", "bedrooms", "bathrooms"]].sum(axis=1)
            availability_den = (z["availability_365"] + 1).where(
                z["availability_365"] >= 0
            )
            z["reviews_per_availability"] = z["number_of_reviews"] / availability_den
            z["neighbourhood_room"] = (
                z["neighbourhood_cleansed"] + "__" + z["room_type"]
            )
            capacity = z["accommodates"].round().astype("int64").astype(str)
            z["property_capacity"] = z["property_type"] + "__cap_" + capacity

        if self.geo:
            geo = z[["latitude", "longitude"]].to_numpy()
            z["geo_cluster"] = self.kmeans_.predict(geo).astype(str)

        numeric = z.select_dtypes(include=np.number).columns
        z[numeric] = z[numeric].replace([np.inf, -np.inf], np.nan).astype(np.float32)
        return z


# Adapter for CatBoost versions with older sklearn estimator tags.
class CatBoostCompat(RegressorMixin, BaseEstimator):
    def __init__(self, estimator):
        self.estimator = estimator

    def fit(self, X, y):
        self.estimator_ = clone(self.estimator).fit(X, y)
        self.n_features_in_ = X.shape[1]
        return self

    def predict(self, X):
        check_is_fitted(self, "estimator_")
        return np.asarray(self.estimator_.predict(X)).reshape(-1)


class PriceRegressor(RegressorMixin, BaseEstimator):
    """Training-only target trimming and optional log1p transformation."""
    def __init__(self, pipeline, log_target=False, upper_quantile=None):
        self.pipeline = pipeline
        self.log_target = log_target
        self.upper_quantile = upper_quantile

    def fit(self, X, y):
        y = np.asarray(y, dtype=float).reshape(-1)
        if len(X) != len(y) or not np.isfinite(y).all() or (y <= 0).any():
            raise ValueError("Training prices must be finite, positive and aligned.")
        q = self.upper_quantile
        if q is not None and not 0 < q <= 1:
            raise ValueError("upper_quantile must be None or in (0, 1].")
        self.cutoff_ = float(np.quantile(y, q)) if q is not None else np.inf
        keep = y <= self.cutoff_
        self.rows_used_ = int(keep.sum())
        self.rows_removed_ = int((~keep).sum())
        self.n_features_in_ = X.shape[1]
        target = np.log1p(y[keep]) if self.log_target else y[keep]
        # Preprocessing is ALSO refitted using only the retained training rows.
        self.pipeline_ = clone(self.pipeline).fit(X.iloc[np.flatnonzero(keep)], target)
        return self

    def predict(self, X):
        check_is_fitted(self, "pipeline_")
        pred = np.asarray(self.pipeline_.predict(X), dtype=float).reshape(-1)
        if self.log_target:
            pred = np.expm1(pred)
        if not np.isfinite(pred).all():
            raise ValueError("Model produced non-finite price predictions.")
        # Same nonnegative rule in CV, holdout and inference; no upper-price cap.
        return np.maximum(pred, 0.0)


def build_price_model(model, setup, log_target=False, upper_quantile=None, scale=False):
    numerical = [("imputer", SimpleImputer(strategy="median", keep_empty_features=True))]
    if scale:
        numerical.append(("scale", StandardScaler()))
    preprocessor = ColumnTransformer([
        ("num", Pipeline(numerical), make_column_selector(dtype_include=np.number)),
        ("cat", OneHotEncoder(
            handle_unknown="ignore", min_frequency=2, max_categories=MAX_CATEGORIES,
            sparse_output=False, dtype=np.float32,
        ), make_column_selector(dtype_exclude=np.number)),
    ], sparse_threshold=0)
    pipeline = Pipeline([
        ("features", AirbnbFeatureEngineer(
            **setup, n_clusters=N_GEO_CLUSTERS, random_state=SEED)),
        ("preprocess", preprocessor),
        ("model", model),
    ])
    return PriceRegressor(pipeline, log_target, upper_quantile)


def make_folds(X, n_splits, seed, by_host):
    if by_host:
        groups = X[GROUP_COL].to_numpy()
        if len(np.unique(groups)) < n_splits:
            raise ValueError("Too few distinct hosts for the requested folds.")
        folds = list(GroupKFold(n_splits=n_splits, shuffle=True, random_state=seed)
                     .split(X, groups=groups))
        for tr, va in folds:
            assert set(groups[tr]).isdisjoint(groups[va])
        return folds
    return list(KFold(n_splits=n_splits, shuffle=True, random_state=seed).split(X))


class GroupAwareStacking(RegressorMixin, BaseEstimator):
    """Every base learner includes its own complete preprocessing pipeline."""
    def __init__(self, estimators, by_host=True, n_splits=3, random_state=2026):
        self.estimators = estimators
        self.by_host = by_host
        self.n_splits = n_splits
        self.random_state = random_state

    def fit(self, X, y):
        # Inner folds are rebuilt on THIS outer training fold, not the full data.
        inner_cv = make_folds(X, self.n_splits, self.random_state, self.by_host)
        self.stack_ = StackingRegressor(
            estimators=self.estimators,
            final_estimator=Pipeline([
                ("scale", StandardScaler()), ("ridge", Ridge(alpha=10.0))]),
            cv=inner_cv, n_jobs=1, passthrough=False,
        ).fit(X, y)
        self.n_features_in_ = X.shape[1]
        return self

    def predict(self, X):
        check_is_fitted(self, "stack_")
        return np.maximum(self.stack_.predict(X), 0.0)


# ------------------------- 3. RAW DATA ----------------------------
started_all = time.perf_counter()
if not Path(DATA_PATH).is_file():
    raise FileNotFoundError(f"Upload the ORIGINAL listings.csv to {DATA_PATH}.")

# Keep large Airbnb IDs as strings; they are not numerical predictors.
df = pd.read_csv(DATA_PATH, dtype={"id": "string", "host_id": "string"})
df.columns = df.columns.str.strip()
if df.columns.duplicated().any():
    raise ValueError("Duplicate CSV column names after stripping whitespace.")
required = RAW_FEATURES + [TARGET] + (["host_id"] if GROUP_BY_HOST else [])
missing = sorted(set(required) - set(df.columns))
if missing:
    raise ValueError(f"Missing columns: {missing}")

n_original = len(df)
df = df.drop_duplicates().copy()
print("Exact duplicate rows removed:", n_original - len(df))
if "id" in df and df["id"].dropna().duplicated().any():
    raise ValueError("Repeated listing IDs remain. Resolve duplicate/snapshot rows first.")

# Accept numeric prices and strings such as '$1,250.00'.
price_text = df[TARGET].astype("string").str.replace(r"[$,\s]", "", regex=True)
y = pd.to_numeric(price_text, errors="coerce").astype(float)
valid = np.isfinite(y) & (y > 0)
print("Rows excluded for missing/invalid/nonpositive price:", int((~valid).sum()))
df, y = df.loc[valid].copy(), y.loc[valid].copy()
if len(df) < 100:
    raise ValueError("At least 100 valid rows are required by this workflow.")
X = df[RAW_FEATURES].copy()
row_ids = df["id"].copy() if "id" in df else pd.Series(df.index, index=df.index)

if GROUP_BY_HOST:
    host = df["host_id"].astype("string").str.strip().replace("", pd.NA)
    fallback = pd.Series([f"missing_host_row:{i}" for i in df.index], index=df.index)
    X[GROUP_COL] = ("host:" + host).fillna(fallback).astype(str)
    tr, te = next(GroupShuffleSplit(n_splits=1, test_size=HOLDOUT_SIZE,
                                   random_state=SEED).split(X, y, X[GROUP_COL]))
    assert set(X.iloc[tr][GROUP_COL]).isdisjoint(X.iloc[te][GROUP_COL])
else:
    tr, te = train_test_split(np.arange(len(X)), test_size=HOLDOUT_SIZE,
                             random_state=SEED)
X_dev, X_holdout = X.iloc[tr].copy(), X.iloc[te].copy()
y_dev, y_holdout = y.iloc[tr].copy(), y.iloc[te].copy()
assert set(X_dev.index).isdisjoint(X_holdout.index)
del df, X, y
gc.collect()

print(f"\nDevelopment: {len(X_dev):,}; holdout: {len(X_holdout):,}")
print("Split:", "non-overlapping hosts" if GROUP_BY_HOST else "random listings")
print("No valid high-price listings have been deleted from validation or holdout.")
print("\nDevelopment missing percentages:")
display((100 * X_dev[RAW_FEATURES].isna().mean()).round(2).to_frame("missing_%"))
print("\nDevelopment price summary:")
display(y_dev.describe(percentiles=[.5, .9, .95, .99, .995]).to_frame("price"))
cv = make_folds(X_dev, CV_FOLDS, SEED + 1, GROUP_BY_HOST)
pd.concat([
    pd.DataFrame({"csv_row_index": X_dev.index, "partition": "development"}),
    pd.DataFrame({"csv_row_index": X_holdout.index, "partition": "holdout"}),
], ignore_index=True).to_csv(OUT / "split_assignments.csv", index=False)

# ---------------------- 4. CV AND MODEL HELPERS -------------------
SCORING = {"MAE": "neg_mean_absolute_error", "RMSE": "neg_root_mean_squared_error", "R2": "r2"}
candidates, records, fold_records, details = {}, [], [], {}


def store_cv(name, estimator, scores, seconds, family, stage, setup_name):
    row = dict(candidate=name, family=family, stage=stage,
               setup=setup_name, seconds=seconds)
    for metric in SCORING:
        sign = 1 if metric == "R2" else -1
        values = sign * np.asarray(scores[f"test_{metric}"])
        row[metric] = float(values.mean())
        row[f"{metric}_std"] = float(values.std(ddof=1))
    row["train_MAE"] = float(-np.mean(scores["train_MAE"]))
    row["MAE_gap"] = row["MAE"] - row["train_MAE"]
    candidates[name] = clone(estimator)
    records.append(row)
    for i in range(CV_FOLDS):
        fold_records.append({"candidate": name, "fold": i + 1, **{
            m: float(scores[f"test_{m}"][i]) * (1 if m == "R2" else -1)
            for m in SCORING}})
    print(f"{name}: CV {PRIMARY_METRIC}={row[PRIMARY_METRIC]:.3f} "
          f"+/- {row[PRIMARY_METRIC + '_std']:.3f}", flush=True)
    pd.DataFrame(records).to_csv(OUT / "model_comparison_cv.csv", index=False)
    pd.DataFrame(fold_records).to_csv(OUT / "fold_scores.csv", index=False)
    return row


def evaluate(name, estimator, family, stage, setup_name):
    print(f"\nCross-validating {name}", flush=True)
    t = time.perf_counter()
    scores = cross_validate(estimator, X_dev, y_dev, cv=cv, scoring=SCORING,
                            return_train_score=True, n_jobs=1, error_score="raise")
    row = store_cv(name, estimator, scores, time.perf_counter() - t,
                   family, stage, setup_name)
    gc.collect()
    return row


def lgbm(**extra):
    params = dict(n_estimators=BOOSTING_ROUNDS, learning_rate=.05,
                  num_leaves=31, min_child_samples=30, subsample=.9,
                  subsample_freq=1, colsample_bytree=.9, reg_lambda=1.0,
                  random_state=SEED, n_jobs=N_JOBS, verbosity=-1,
                  deterministic=True, force_col_wise=True)
    params.update(extra)
    return LGBMRegressor(**params)

# -------------------- 5. FEATURE/TARGET ABLATION ------------------
setups = {
    "A_original": dict(engineered=False, geo=False, missing_flags=False),
    "B_engineered": dict(engineered=True, geo=False, missing_flags=False),
    "C_engineered_geo": dict(engineered=True, geo=True, missing_flags=False),
    "D_engineered_geo_missing": dict(engineered=True, geo=True, missing_flags=True),
}
# Same LGBM parameters and exact validation rows for every experiment.
for setup_name, setup in setups.items():
    for log in [False, True]:
        name = f"LightGBM__{setup_name}__{'log' if log else 'raw'}"
        evaluate(name, build_price_model(lgbm(), setup, log),
                 "LightGBM", "feature_target_ablation", setup_name)

ablation = pd.DataFrame(records).sort_values(PRIMARY_METRIC)
ablation.to_csv(OUT / "feature_target_ablation.csv", index=False)
print("\nFEATURE/TARGET ABLATION — development scores, not independent test scores")
display(ablation.round(4))
best_feature_row = ablation.iloc[0]
best_setup_name = best_feature_row["setup"]
best_setup = setups[best_setup_name]
reference = candidates[best_feature_row["candidate"]]

# ------------------- 6. TRAINING-ONLY OUTLIER TEST -----------------
# Quantiles are recalculated INSIDE each training fold.
# Validation targets are NOT used to compute or apply the cutoff.
# All candidates see the SAME complete validation folds.
for q in [.99, .995]:
    evaluate(f"LightGBM__training_trim_{q}",
             clone(reference).set_params(upper_quantile=q),
             "LightGBM", "training_outlier_ablation", best_setup_name)

trim_rows = [r for r in records if r["candidate"] == best_feature_row["candidate"]
             or r["stage"] == "training_outlier_ablation"]
trim_table = pd.DataFrame(trim_rows).sort_values(PRIMARY_METRIC)
trim_table.to_csv(OUT / "training_outlier_ablation.csv", index=False)
display(trim_table.round(4))
best_trim = candidates[trim_table.iloc[0]["candidate"]].upper_quantile

# ------------------------- 7. MODEL COMPARISON --------------------
# Shared feature setup selected above; not a separate exhaustive
# feature search for each algorithm. Raw/log target tested for each.
models = {
    "LinearRegression": LinearRegression(),
    "Ridge": Ridge(alpha=10.0),
    "RandomForest": RandomForestRegressor(
        n_estimators=RF_TREES, max_depth=22, min_samples_leaf=2,
        max_features=.8, n_jobs=N_JOBS, random_state=SEED),
    "XGBoost": XGBRegressor(
        n_estimators=BOOSTING_ROUNDS, learning_rate=.05, max_depth=6,
        min_child_weight=3, subsample=.9, colsample_bytree=.9,
        objective="reg:squarederror", tree_method="hist",
        n_jobs=N_JOBS, random_state=SEED),
    "CatBoost": CatBoostCompat(CatBoostRegressor(
        iterations=BOOSTING_ROUNDS, learning_rate=.05, depth=6,
        loss_function="RMSE", thread_count=N_JOBS, random_seed=SEED,
        verbose=False, allow_writing_files=False)),
}
for family, model in models.items():
    for log in [False, True]:
        name = f"{family}__{'log' if log else 'raw'}"
        evaluate(name, build_price_model(clone(model), best_setup, log, best_trim,
                  scale=family in {"LinearRegression", "Ridge"}),
                 family, "model_comparison", best_setup_name)

evaluate("Median_baseline", build_price_model(DummyRegressor(strategy="median"),
         setups["A_original"]), "Baseline", "baseline", "A_original")

# ------------------- 8. TUNE THREE BOOSTING MODELS ----------------
spaces = {
    "LightGBM": {
        "n_estimators": [300, 600, 900], "learning_rate": [.025, .05, .08],
        "num_leaves": [15, 31, 63], "max_depth": [-1, 6, 8],
        "min_child_samples": [20, 40, 80], "reg_alpha": [0.0, .5, 2.0],
        "reg_lambda": [.5, 2.0, 5.0],
    },
    "XGBoost": {
        "n_estimators": [300, 600, 900], "learning_rate": [.025, .05, .08],
        "max_depth": [4, 6, 8], "min_child_weight": [1, 3, 8],
        "subsample": [.75, .9, 1.0], "colsample_bytree": [.75, .9, 1.0],
        "reg_lambda": [1.0, 3.0, 10.0],
    },
    "CatBoost": {
        "iterations": [300, 600, 900], "learning_rate": [.025, .05, .08],
        "depth": [4, 6, 8], "l2_leaf_reg": [1.0, 3.0, 7.0, 10.0],
        "random_strength": [0.0, .5, 1.0],
    },
}
for family, space in spaces.items():
    eligible = [r for r in records if r["family"] == family]
    previous_best = min(eligible, key=lambda r: r[PRIMARY_METRIC])
    template = clone(candidates[previous_best["candidate"]])
    prefix = "pipeline__model__" + ("estimator__" if family == "CatBoost" else "")
    params = {prefix + k: v for k, v in space.items()}
    params.update(log_target=[False, True], upper_quantile=[None, .99, .995])
    print(f"\nTUNING {family}", flush=True)
    t = time.perf_counter()
    search = RandomizedSearchCV(template, params, n_iter=SEARCH_ITERATIONS,
             scoring=SCORING, cv=cv, refit=False, n_jobs=1, verbose=1,
             random_state=SEED + 2, return_train_score=True, error_score="raise")
    search.fit(X_dev, y_dev)
    res = search.cv_results_
    pd.DataFrame(res).to_csv(OUT / f"{family}_search.csv", index=False)
    best_idx = int(np.argmax(res[f"mean_test_{PRIMARY_METRIC}"]))
    best_params = res["params"][best_idx]
    tuned = clone(template).set_params(**best_params)
    scores = {f"{split}_{metric}": np.array([
        res[f"split{i}_{split}_{metric}"][best_idx] for i in range(CV_FOLDS)
    ]) for split in ["test", "train"] for metric in SCORING}
    details[family] = best_params
    print("Best search parameters:", best_params)
    store_cv(f"{family}__tuned", tuned, scores, time.perf_counter() - t,
             family, "hyperparameter_search", previous_best["setup"])
    del search
    gc.collect()

# ------------------- 9. VOTING AND GROUP-AWARE STACKING ------------
# Each base estimator contains its own features, imputation and encoding.
# No outer shared transformer fitted before stacking's inner folds.
ensemble_bases = []
for family in ["LightGBM", "XGBoost", "CatBoost", "RandomForest"]:
    best = min((r for r in records if r["family"] == family),
               key=lambda r: r[PRIMARY_METRIC])
    ensemble_bases.append((family.lower(), clone(candidates[best["candidate"]])))
    print("Ensemble component:", best["candidate"])

if RUN_VOTING:
    evaluate("Voting_equal_weights", VotingRegressor(ensemble_bases, n_jobs=1),
             "Ensemble", "voting", "component-specific")
if RUN_STACKING:
    evaluate("Stacking", GroupAwareStacking(ensemble_bases, by_host=GROUP_BY_HOST,
             n_splits=CV_FOLDS, random_state=SEED + 3),
             "Ensemble", "stacking", "component-specific")

# ------------------- 10. FREEZE WINNER AND REFIT -------------------
leaderboard = pd.DataFrame(records).sort_values(
    [PRIMARY_METRIC, "seconds"], kind="stable").reset_index(drop=True)
leaderboard.to_csv(OUT / "model_comparison_cv.csv", index=False)
print("\nFINAL DEVELOPMENT LEADERBOARD — selection scores, not nested-CV estimates")
display(leaderboard.round(4))
winner_name = leaderboard.iloc[0]["candidate"]
final_model = clone(candidates[winner_name])
print(f"\nFitting selected model on all development data: {winner_name}", flush=True)
final_model.fit(X_dev, y_dev)
# A selected trimming policy may remove TRAINING rows during this fit.
# The holdout remains complete. No model choice follows holdout evaluation.

# ----------------------- 11. FINAL HOLDOUT RESULTS ----------------
def metrics(y_true, pred):
    a, p = np.asarray(y_true, float), np.asarray(pred, float)
    return {
        "rows": len(a), "MAE": mean_absolute_error(a, p),
        "RMSE": np.sqrt(mean_squared_error(a, p)),
        "R2": r2_score(a, p) if len(a) > 1 else np.nan,
        "RMSLE": np.sqrt(mean_squared_log_error(a, p)),
        "median_absolute_error": median_absolute_error(a, p),
        "WAPE_percent": 100 * np.abs(a - p).sum() / np.abs(a).sum(),
        "mean_residual_actual_minus_pred": float(np.mean(a - p)),
    }

pred = final_model.predict(X_holdout)
baseline_pred = np.full(len(y_holdout), y_dev.median())
final_results = pd.DataFrame([
    {"model": "Development-median baseline", **metrics(y_holdout, baseline_pred)},
    {"model": winner_name, **metrics(y_holdout, pred)},
])
print("\nFINAL HOLDOUT — ALL VALID PRICES, including the expensive tail")
display(final_results.round(4))
final_results.to_csv(OUT / "final_holdout_metrics.csv", index=False)
base_mae, final_mae = final_results["MAE"].to_numpy()
print(f"MAE improvement over baseline: {base_mae - final_mae:.3f} "
      f"({100 * (base_mae - final_mae) / base_mae:.2f}%)" if base_mae > 0
      else "Baseline MAE is zero; percentage improvement is undefined.")

# -------------------- 12. SAVE EXACT EVALUATED MODEL --------------
versions = {p: metadata.version(p) for p in minimum_versions}
versions.update({p: metadata.version(p) for p in ["scipy", "threadpoolctl"]})
bundle = {"model": final_model, "raw_features": RAW_FEATURES, "target": TARGET,
          "winner": winner_name, "selection_metric": PRIMARY_METRIC,
          "group_by_host": GROUP_BY_HOST, "seed": SEED, "versions": versions,
          "development_rows": len(X_dev), "holdout_rows": len(X_holdout),
          "best_search_parameters": details, "python_version": sys.version,
          "configuration": {"cv_folds": CV_FOLDS, "search_iterations": SEARCH_ITERATIONS,
                            "holdout_size": HOLDOUT_SIZE, "geo_clusters": N_GEO_CLUSTERS,
                            "max_categories": MAX_CATEGORIES, "boosting_rounds": BOOSTING_ROUNDS,
                            "rf_trees": RF_TREES, "voting": RUN_VOTING, "stacking": RUN_STACKING}}
with open(OUT / "model_bundle.pkl", "wb") as f:
    cloudpickle.dump(bundle, f)
with open(OUT / "run_metadata.json", "w") as f:
    json.dump({k: v for k, v in bundle.items() if k != "model"}, f,
              indent=2, default=lambda x: x.item() if hasattr(x, "item") else str(x))
(OUT / "requirements.txt").write_text("\n".join(f"{k}=={v}" for k, v in versions.items()) + "\n")


def predict_new(raw_data, model_bundle=bundle):
    if isinstance(raw_data, dict):
        raw_data = pd.DataFrame([raw_data])
    elif not isinstance(raw_data, pd.DataFrame):
        raw_data = pd.DataFrame(raw_data)
    # host_id and __host_group are NOT required at prediction time.
    return pd.DataFrame({"predicted_price": model_bundle["model"].predict(raw_data)},
                        index=raw_data.index)

with open(OUT / "model_bundle.pkl", "rb") as f:
    reloaded = cloudpickle.load(f)
sample = X_holdout[RAW_FEATURES].head(5)
np.testing.assert_allclose(predict_new(sample), predict_new(sample, reloaded))
print("\nSaved/reloaded model prediction example:")
display(predict_new(sample))

# Save feature names and training cutoff information for each component.
def fitted_components(model):
    if isinstance(model, PriceRegressor):
        return [("model", model)]
    ensemble = model.stack_ if isinstance(model, GroupAwareStacking) else model
    return list(zip([name for name, _ in ensemble.estimators], ensemble.estimators_))

feature_rows, training_rows = [], []
for name, fitted in fitted_components(final_model):
    names = fitted.pipeline_.named_steps["preprocess"].get_feature_names_out()
    feature_rows.extend({"component": name, "feature": n} for n in names)
    training_rows.append({"component": name, "log_target": fitted.log_target,
                         "upper_quantile": fitted.upper_quantile,
                         "cutoff": fitted.cutoff_, "rows_used": fitted.rows_used_,
                         "rows_removed": fitted.rows_removed_})
pd.DataFrame(feature_rows).to_csv(OUT / "final_feature_names.csv", index=False)
pd.DataFrame(training_rows).to_csv(OUT / "final_training_policies.csv", index=False)
print("\nFinal component training policies:")
display(pd.DataFrame(training_rows))

# --------------------- 13. RESIDUAL DIAGNOSTICS -------------------
# Descriptive only: do NOT change this model after seeing these results.
diag = X_holdout[RAW_FEATURES].copy()
diag.insert(0, "id", row_ids.loc[diag.index])
diag["actual_price"], diag["predicted_price"] = y_holdout, pred
diag["residual"] = diag["actual_price"] - diag["predicted_price"]
diag["absolute_error"] = diag["residual"].abs()
diag["price_band"] = pd.cut(diag["actual_price"],
    bins=[0, 100, 200, 500, 1000, 2000, np.inf],
    labels=["0-100", "100-200", "200-500", "500-1000", "1000-2000", "2000+"])
diag["missing_count"] = X_holdout[RAW_FEATURES].isna().sum(axis=1)
diag.to_csv(OUT / "holdout_predictions.csv", index_label="csv_row_index")

for column in ["price_band", "room_type", "neighbourhood_cleansed", "missing_count"]:
    table = diag.groupby(column, observed=True, dropna=False).agg(
        rows=("absolute_error", "size"), MAE=("absolute_error", "mean"),
        median_error=("absolute_error", "median"),
        mean_actual=("actual_price", "mean"), mean_residual=("residual", "mean"))
    table.to_csv(OUT / f"errors_by_{column}.csv")
    print(f"\nErrors by {column}:")
    shown = table[table["rows"] >= 10] if column == "neighbourhood_cleansed" else table
    display(shown.sort_values("MAE", ascending=False).head(20).round(3))

# Optional descriptive tail comparison; headline scores above still use ALL rows.
tail_cutoff = float(y_dev.quantile(.99))
tail_results = []
for label, mask in [("At/below development 99th percentile", y_holdout <= tail_cutoff),
                    ("Above development 99th percentile", y_holdout > tail_cutoff)]:
    if mask.any():
        tail_results.append({"segment": label, "cutoff": tail_cutoff,
                             **metrics(y_holdout.loc[mask], pred[mask.to_numpy()])})
pd.DataFrame(tail_results).to_csv(OUT / "holdout_tail_diagnostics.csv", index=False)
display(pd.DataFrame(tail_results).round(3))
display(diag.nlargest(15, "absolute_error")[[
    "actual_price", "predicted_price", "absolute_error", "neighbourhood_cleansed",
    "room_type", "accommodates"]])

# All holdout listings are represented in these plots.
plt.figure(figsize=(7, 6))
plt.scatter(y_holdout, pred, s=10, alpha=.35)
limit = max(float(y_holdout.max()), float(np.max(pred)))
plt.plot([0, limit], [0, limit], linestyle="--")
plt.xscale("symlog", linthresh=1); plt.yscale("symlog", linthresh=1)
plt.xlabel("Actual price (symlog scale)"); plt.ylabel("Predicted price (symlog scale)")
plt.title("Holdout actual versus predicted — complete price range")
plt.tight_layout(); plt.savefig(OUT / "actual_vs_predicted.png", dpi=150); plt.show()

plt.figure(figsize=(8, 5))
plt.scatter(pred, diag["residual"], s=10, alpha=.35)
plt.axhline(0, linestyle="--")
plt.xscale("symlog", linthresh=1)
plt.xlabel("Predicted price (symlog scale)"); plt.ylabel("Actual minus predicted price")
plt.title("Holdout residuals — no tail exclusions")
plt.tight_layout(); plt.savefig(OUT / "residuals.png", dpi=150); plt.show()

# ------------------------ 14. PERMUTATION IMPORTANCE --------------
# Post-evaluation interpretation only; NEVER used to reselect features.
if RUN_PERMUTATION:
    X_perm = X_holdout[RAW_FEATURES].sample(
        min(PERMUTATION_ROWS, len(X_holdout)), random_state=SEED)
    y_perm = y_holdout.loc[X_perm.index]
    print("\nComputing descriptive holdout permutation importance...", flush=True)
    result = permutation_importance(final_model, X_perm, y_perm,
        scoring="neg_mean_absolute_error", n_repeats=PERMUTATION_REPEATS,
        random_state=SEED, n_jobs=1)
    importance = pd.DataFrame({"raw_feature": RAW_FEATURES,
        "MAE_increase": result.importances_mean,
        "std": result.importances_std}).sort_values("MAE_increase", ascending=False)
    importance.to_csv(OUT / "permutation_importance.csv", index=False)
    display(importance.round(4))
    plot = importance.sort_values("MAE_increase")
    plt.figure(figsize=(9, 6))
    plt.barh(plot["raw_feature"], plot["MAE_increase"], xerr=plot["std"])
    plt.xlabel("Increase in MAE after shuffling raw input")
    plt.title("Raw-feature importance; engineered features recomputed")
    plt.tight_layout(); plt.savefig(OUT / "permutation_importance.png", dpi=150); plt.show()

# ----------------------------- 15. DONE --------------------------
print("\n" + "=" * 70)
print("SELECTED MODEL:", winner_name)
print("ALL-HOLDOUT MAE:", round(float(final_results.iloc[1]["MAE"]), 4))
print("ALL-HOLDOUT RMSE:", round(float(final_results.iloc[1]["RMSE"]), 4))
print("ALL-HOLDOUT R2:", round(float(final_results.iloc[1]["R2"]), 4))
print("Saved outputs:", OUT)
print("New raw-row predictions: predict_new(your_dataframe)")
print(f"Actual elapsed time: {time.perf_counter() - started_all:.1f} seconds")
print("This is a holdout for this run, not a brand-new external dataset.")
print("Listed-price prediction is not a forecast of bookings or earned revenue.")

# LATER, in a fresh session with the saved dependency versions:
# import cloudpickle
# with open('/content/airbnb_results/model_bundle.pkl', 'rb') as f:
#     saved = cloudpickle.load(f)  # Load ONLY trusted pickle files.
# predicted_prices = saved['model'].predict(new_raw_dataframe)
```

```python
# ================================================================
# AIRBNB V2 — same holdout policy, MAE-focused candidate search
# Run in ONE fresh Colab cell with the SAME /content/listings.csv.
# No valid expensive listings are removed from training or holdout.
# This is an improvement experiment, NOT a guaranteed score increase.
# ================================================================

# -------------------------- 1. SETTINGS ---------------------------
DATA_PATH = "/content/listings.csv"
OUTPUT_DIR = "/content/airbnb_results_v2"
PREVIOUS_DIR = "/content/airbnb_results"  # Set None for an unrelated CSV.
SEED = 2026
GROUP_BY_HOST = True                    # Same as the previous snippet.
HOLDOUT_SIZE = 0.20
BLEND_VALIDATION_SIZE = 0.20             # Of DEVELOPMENT hosts, not holdout.
CV_FOLDS = 3
TRIALS_PER_FAMILY = 8                   # Includes four fixed starting trials.
TOP_MODELS_PER_FAMILY = 2
TRY_EXTRA_FEATURES = True               # Tested, not automatically retained.
BLEND_STEPS = 50
MIN_BLEND_GAIN = 0.25                   # Validation MAE gain vs best single.

# Fresh runtime recommended; installs only missing/too-old dependencies.
import os, sys, gc, json, time, hashlib, subprocess, warnings
import importlib.metadata as metadata
from pathlib import Path
from copy import deepcopy
from packaging.version import Version

minimum = {
    "numpy": "1.24", "pandas": "2.0", "scikit-learn": "1.6",
    "lightgbm": "4.6.0", "catboost": "1.2.8", "xgboost": "3.1.3",
    "cloudpickle": "3.1", "matplotlib": "3.7",
}
needed = []
for package, version in minimum.items():
    try:
        installed = metadata.version(package)
    except metadata.PackageNotFoundError:
        installed = "0"
    if Version(installed) < Version(version):
        needed.append(f"{package}>={version}")
if needed:
    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "-q", *needed]
    )

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cloudpickle
from IPython.display import display
from sklearn.model_selection import (
    GroupShuffleSplit, GroupKFold, KFold, train_test_split
)
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score
)
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
from xgboost import XGBRegressor
from threadpoolctl import threadpool_limits

N_JOBS = max(1, min(4, os.cpu_count() or 1))
_thread_limit = threadpool_limits(limits=N_JOBS)

warnings.filterwarnings(
    "ignore",
    message=(
        "X does not have valid feature names, but "
        "LGBMRegressor was fitted with feature names"
    ),
)

OUT = Path(OUTPUT_DIR)
OUT.mkdir(parents=True, exist_ok=True)

if (
    CV_FOLDS < 2
    or TRIALS_PER_FAMILY < 4
    or TOP_MODELS_PER_FAMILY < 1
):
    raise ValueError(
        "Use CV_FOLDS>=2, TRIALS_PER_FAMILY>=4 "
        "and TOP_MODELS_PER_FAMILY>=1."
    )

if (
    PREVIOUS_DIR
    and OUT.resolve() == Path(PREVIOUS_DIR).resolve()
):
    raise ValueError(
        "Use a NEW output directory; preserve the previous results."
    )

NUM = [
    "latitude", "longitude", "accommodates", "bathrooms",
    "bedrooms", "beds", "minimum_nights", "availability_365",
    "number_of_reviews",
]
CAT = [
    "neighbourhood_cleansed", "property_type", "room_type"
]
CORE = NUM + CAT

# Explicit optional allowlist.
# Never use price-derived revenue/occupancy fields, host/listing IDs,
# URLs, or an unrestricted set of CSV columns as predictors.
EXTRA_NUM = [
    "host_response_rate", "host_listings_count"
]
EXTRA_CAT = [
    "host_response_time", "host_is_superhost",
    "instant_bookable", "neighbourhood_group_cleansed",
]
EXTRA_TEXT = [
    "amenities", "bathrooms_text"
]


# -------------------- 2. RAW-FEATURE TRANSFORMER ------------------
class FeatureTable:
    def __init__(self, mode="lgb", rich=False):
        self.mode, self.rich = mode, rich

    def _make(self, X):
        if not isinstance(X, pd.DataFrame):
            raise TypeError(
                "Pass raw inputs as a pandas DataFrame."
            )

        missing = sorted(set(self.inputs_) - set(X.columns))
        if missing:
            raise ValueError(
                f"Missing raw input columns: {missing}"
            )

        z = pd.DataFrame(index=X.index)

        for c in self.numeric_inputs_:
            values = X[c]
            if c == "host_response_rate":
                values = (
                    values.astype("string")
                    .str.replace("%", "", regex=False)
                )
            z[c] = pd.to_numeric(
                values, errors="coerce"
            ).astype(float)

        z = z.replace([np.inf, -np.inf], np.nan)

        z["latitude"] = z["latitude"].where(
            z["latitude"].between(-90, 90)
        )
        z["longitude"] = z["longitude"].where(
            z["longitude"].between(-180, 180)
        )

        for c in self.category_inputs_:
            z[c] = (
                X[c].astype("string")
                .str.strip()
                .replace("", pd.NA)
                .fillna("Unknown")
                .astype(object)
            )

        if self.rich:
            for c in self.numeric_inputs_:
                z[c + "__missing"] = (
                    z[c].isna().astype(np.float32)
                )

            denom = z["accommodates"].where(
                z["accommodates"] > 0
            )

            for c in ["beds", "bedrooms", "bathrooms"]:
                z[c + "_per_guest"] = z[c] / denom

            z["neighbourhood_room"] = (
                z[CAT[0]] + "__" + z["room_type"]
            )
            z["property_room"] = (
                z["property_type"] + "__" + z["room_type"]
            )
            z["geo_cell"] = (
                z["latitude"].round(2).astype(str)
                + "__"
                + z["longitude"].round(2).astype(str)
            )

            if "amenities" in self.inputs_:
                text = (
                    X["amenities"].astype("string").str.lower()
                )
                z["amenities_missing"] = (
                    text.isna().astype(np.float32)
                )

                # Text counts/flags, not a learned vocabulary
                # or target encoding.
                z["amenity_separator_count"] = (
                    text.str.count(",").astype(float)
                )

                patterns = {
                    "pool": r"\bpool\b",
                    "hot_tub": r"hot tub|jacuzzi",
                    "parking": r"parking",
                    "air_conditioning": r"air conditioning",
                    "kitchen": r"kitchen",
                    "washer": r"washer|washing machine",
                    "wifi": r"wifi|wi-fi",
                    "gym": r"gym",
                    "fireplace": r"fireplace",
                    "waterfront": (
                        r"waterfront|beachfront|beach access"
                    ),
                }

                for name, pattern in patterns.items():
                    z["amenity_" + name] = (
                        text.str.contains(
                            pattern, regex=True, na=False
                        ).astype(np.float32)
                    )

            if "bathrooms_text" in self.inputs_:
                text = (
                    X["bathrooms_text"]
                    .astype("string")
                    .str.lower()
                )
                z["shared_bathroom"] = (
                    text.str.contains(
                        "shared", na=False
                    ).astype(float)
                )

        numeric = z.select_dtypes(
            include=np.number
        ).columns

        z[numeric] = (
            z[numeric]
            .replace([np.inf, -np.inf], np.nan)
            .astype(np.float32)
        )

        return z

    def fit(self, X):
        self.numeric_inputs_ = NUM + (
            [c for c in EXTRA_NUM if c in X]
            if self.rich else []
        )
        self.category_inputs_ = CAT + (
            [c for c in EXTRA_CAT if c in X]
            if self.rich else []
        )

        self.inputs_ = (
            self.numeric_inputs_ + self.category_inputs_
        )

        if self.rich:
            self.inputs_ += [
                c for c in EXTRA_TEXT if c in X
            ]

        z = self._make(X)
        self.columns_ = list(z.columns)

        self.cat_columns_ = (
            z.select_dtypes(exclude=np.number)
            .columns.tolist()
        )
        self.num_columns_ = (
            z.select_dtypes(include=np.number)
            .columns.tolist()
        )

        if self.mode == "ohe":
            # Reproduces the previous baseline's
            # median/one-hot recipe.
            self.encoder_ = ColumnTransformer(
                [
                    (
                        "num",
                        SimpleImputer(
                            strategy="median",
                            keep_empty_features=True,
                        ),
                        self.num_columns_,
                    ),
                    (
                        "cat",
                        OneHotEncoder(
                            handle_unknown="ignore",
                            min_frequency=2,
                            max_categories=128,
                            sparse_output=False,
                            dtype=np.float32,
                        ),
                        self.cat_columns_,
                    ),
                ],
                sparse_threshold=0,
            ).fit(z)

        elif self.mode == "lgb":
            self.levels_ = {
                c: sorted(z[c].unique().tolist())
                for c in self.cat_columns_
            }

        return self

    def transform(self, X):
        z = self._make(X)[self.columns_]

        if self.mode == "ohe":
            return self.encoder_.transform(z)

        if self.mode == "lgb":
            # Training-fold category mapping.
            # Unseen categories become missing.
            for c in self.cat_columns_:
                z[c] = pd.Categorical(
                    z[c], categories=self.levels_[c]
                )

        return z


class PriceModel:
    def __init__(self, spec):
        self.spec = deepcopy(spec)

    def fit(self, X, y):
        s = self.spec
        family = s["family"]

        self.features_ = FeatureTable(
            s["encoding"],
            s.get("rich", False),
        ).fit(X)

        train = self.features_.transform(X)
        target = np.asarray(y, dtype=float)

        if s["log"]:
            target = np.log1p(target)

        p = deepcopy(s["params"])

        if family == "lgb":
            self.model_ = LGBMRegressor(
                random_state=SEED,
                n_jobs=N_JOBS,
                verbosity=-1,
                deterministic=True,
                force_col_wise=True,
                **p,
            )
            self.model_.fit(train, target)

        elif family == "cat":
            self.model_ = CatBoostRegressor(
                random_seed=SEED,
                thread_count=N_JOBS,
                verbose=False,
                allow_writing_files=False,
                **p,
            )

            cats = (
                self.features_.cat_columns_
                if s["encoding"] == "cat"
                else None
            )
            self.model_.fit(
                train, target, cat_features=cats
            )

        elif family == "xgb":
            self.model_ = XGBRegressor(
                random_state=SEED,
                n_jobs=N_JOBS,
                tree_method="hist",
                objective="reg:squarederror",
                **p,
            ).fit(train, target)

        elif family == "rf":
            self.model_ = RandomForestRegressor(
                random_state=SEED,
                n_jobs=N_JOBS,
                **p,
            ).fit(train, target)

        else:
            raise ValueError(
                f"Unknown model family: {family}"
            )

        return self

    def predict(self, X):
        pred = np.asarray(
            self.model_.predict(
                self.features_.transform(X)
            ),
            float,
        )

        if self.spec["log"]:
            pred = np.expm1(pred)

        if not np.isfinite(pred).all():
            raise ValueError(
                "Non-finite model predictions."
            )

        # Same nonnegative rule for every model.
        return np.maximum(pred, 0.0)


class BlendModel:
    def __init__(self, models, weights):
        self.models, self.weights = models, weights

    def predict(self, X):
        return sum(
            self.weights[n] * self.models[n].predict(X)
            for n in self.weights
            if self.weights[n] > 0
        )


def metrics(y, p):
    return {
        "MAE": mean_absolute_error(y, p),
        "RMSE": np.sqrt(mean_squared_error(y, p)),
        "R2": r2_score(y, p),
    }


def split_positions(X, groups, fraction, seed):
    if GROUP_BY_HOST:
        return next(
            GroupShuffleSplit(
                n_splits=1,
                test_size=fraction,
                random_state=seed,
            ).split(X, groups=groups)
        )

    return train_test_split(
        np.arange(len(X)),
        test_size=fraction,
        random_state=seed,
    )


# ----------------------- 3. SAME RAW ROW POLICY -------------------
started = time.perf_counter()

if not Path(DATA_PATH).is_file():
    raise FileNotFoundError(
        f"Upload the SAME original CSV to {DATA_PATH}."
    )

raw = pd.read_csv(
    DATA_PATH,
    dtype={"id": "string", "host_id": "string"},
    low_memory=False,
)
raw.columns = raw.columns.str.strip()

if raw.columns.duplicated().any():
    raise ValueError("Duplicate CSV column names.")

required = (
    CORE + ["price"]
    + (["host_id"] if GROUP_BY_HOST else [])
)

if set(required) - set(raw.columns):
    raise ValueError(
        f"Missing columns: "
        f"{sorted(set(required) - set(raw.columns))}"
    )

original_rows = len(raw)
raw = raw.drop_duplicates().copy()

print(
    "Rows loaded:", original_rows,
    "| exact duplicates removed:",
    original_rows - len(raw),
)

if (
    "id" in raw
    and raw["id"].dropna().duplicated().any()
):
    raise ValueError(
        "Repeated listing IDs remain; "
        "resolve repeated snapshots first."
    )

y = pd.to_numeric(
    raw["price"].astype("string").str.replace(
        r"[$,\s]", "", regex=True
    ),
    errors="coerce",
).astype(float)

valid = np.isfinite(y) & (y > 0)

print(
    "Invalid/nonpositive price rows removed:",
    int((~valid).sum()),
)

raw = raw.loc[valid].copy()
y = y.loc[valid].copy()

if len(raw) < 200:
    raise ValueError(
        "This workflow expects at least 200 valid listings."
    )

ids = (
    raw["id"].copy()
    if "id" in raw
    else pd.Series(
        raw.index, index=raw.index
    ).astype(str)
)

if GROUP_BY_HOST:
    host = (
        raw["host_id"]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
    )
    fallback = pd.Series(
        [f"missing_host_row:{i}" for i in raw.index],
        index=raw.index,
    )
    groups = (
        ("host:" + host)
        .fillna(fallback)
        .astype(str)
    )
else:
    groups = pd.Series(
        raw.index.astype(str),
        index=raw.index,
    )

extras = [
    c for c in EXTRA_NUM + EXTRA_CAT + EXTRA_TEXT
    if c in raw
]
X = raw[CORE + extras].copy()
del raw

# Reuse saved assignments when available.
# A changed CSV must not silently produce a "same-holdout"
# comparison based only on a matching random seed.
prior_split = (
    Path(PREVIOUS_DIR) / "split_assignments.csv"
    if PREVIOUS_DIR else None
)

if prior_split and prior_split.exists():
    saved = pd.read_csv(prior_split)
    expected_cols = {"csv_row_index", "partition"}

    if (
        not expected_cols.issubset(saved.columns)
        or saved["csv_row_index"].duplicated().any()
    ):
        raise ValueError("Invalid saved split file.")

    if set(saved["csv_row_index"]) != set(X.index):
        raise ValueError(
            "CSV rows differ from the saved run. "
            "Use the original CSV, or set PREVIOUS_DIR=None "
            "for a new, nonhistorical comparison."
        )

    if set(saved["partition"]) != {
        "development", "holdout"
    }:
        raise ValueError(
            "Unexpected partition names in saved split."
        )

    dev_idx = saved.loc[
        saved.partition == "development",
        "csv_row_index",
    ].to_numpy()

    test_idx = saved.loc[
        saved.partition == "holdout",
        "csv_row_index",
    ].to_numpy()

    print("Reusing previous row assignments.")

else:
    dev_pos, test_pos = split_positions(
        X, groups, HOLDOUT_SIZE, SEED
    )
    dev_idx = X.index[dev_pos]
    test_idx = X.index[test_pos]

    print(
        "Recreating the previous seed/split policy; "
        "no saved assignments found."
    )

Xd = X.loc[dev_idx]
Xt = X.loc[test_idx]
yd = y.loc[dev_idx]
yt = y.loc[test_idx]
gd = groups.loc[dev_idx]
gt = groups.loc[test_idx]

assert set(Xd.index).isdisjoint(Xt.index)

if GROUP_BY_HOST:
    if not set(gd).isdisjoint(gt):
        raise ValueError(
            "Saved split overlaps hosts; "
            "it is not a host-grouped evaluation."
        )

pd.concat(
    [
        pd.DataFrame(
            {
                "csv_row_index": Xd.index,
                "partition": "development",
            }
        ),
        pd.DataFrame(
            {
                "csv_row_index": Xt.index,
                "partition": "holdout",
            }
        ),
    ]
).to_csv(
    OUT / "split_assignments.csv",
    index=False,
)

# Separate development-only data for selecting ensemble weights.
fit_pos, blend_pos = split_positions(
    Xd, gd, BLEND_VALIDATION_SIZE, SEED + 31
)

Xfit = Xd.iloc[fit_pos]
Xblend = Xd.iloc[blend_pos]
yfit = yd.iloc[fit_pos]
yblend = yd.iloc[blend_pos]
gfit = gd.iloc[fit_pos]
gblend = gd.iloc[blend_pos]

if GROUP_BY_HOST:
    assert set(gfit).isdisjoint(gblend)

    cv = list(
        GroupKFold(
            n_splits=CV_FOLDS,
            shuffle=True,
            random_state=SEED + 1,
        ).split(Xfit, groups=gfit)
    )

    for tr, va in cv:
        assert set(gfit.iloc[tr]).isdisjoint(
            gfit.iloc[va]
        )
else:
    cv = list(
        KFold(
            n_splits=CV_FOLDS,
            shuffle=True,
            random_state=SEED + 1,
        ).split(Xfit)
    )

print(
    f"\nDevelopment={len(Xd):,}; "
    f"final holdout={len(Xt):,}"
)
print(
    f"Within development: CV/training={len(Xfit):,}; "
    f"blend validation={len(Xblend):,}"
)
print("Optional allowlisted inputs available:", extras)
print(
    "No price-tail trimming. "
    "IDs and host IDs are never predictors."
)


# -------------------- 4. TARGETED NATIVE-BOOSTER SEARCH ------------
# Fixed tree counts are tuned like other parameters.
# There is NO early stopping on the fold being scored,
# and NO model selection on holdout.

rng = np.random.default_rng(SEED)
configs = {}

for family in ["lgb", "cat"]:
    for trial in range(TRIALS_PER_FAMILY):
        if trial < 4:
            log = trial % 2 == 0
            rich = bool(
                TRY_EXTRA_FEATURES and trial >= 2
            )
        else:
            log = bool(rng.integers(2))
            rich = bool(
                TRY_EXTRA_FEATURES and rng.integers(2)
            )

        if family == "lgb":
            p = dict(
                objective=(
                    "regression" if log
                    else "regression_l1"
                ),
                n_estimators=900,
                learning_rate=.035,
                num_leaves=31,
                min_child_samples=30,
                reg_lambda=2.0,
                reg_alpha=0.0,
                subsample=.9,
                subsample_freq=1,
                colsample_bytree=.9,
                cat_smooth=20.0,
                cat_l2=10.0,
            )

            if trial >= 4:
                p.update(
                    n_estimators=int(
                        rng.choice([600, 1000, 1600])
                    ),
                    learning_rate=float(
                        rng.choice([.025, .04, .06])
                    ),
                    num_leaves=int(
                        rng.choice([15, 31, 63])
                    ),
                    min_child_samples=int(
                        rng.choice([15, 30, 60, 100])
                    ),
                    reg_lambda=float(
                        rng.choice([1.0, 3.0, 8.0])
                    ),
                    cat_smooth=float(
                        rng.choice([5.0, 20.0, 50.0])
                    ),
                )
        else:
            p = dict(
                loss_function="RMSE" if log else "MAE",
                iterations=1000,
                learning_rate=.035,
                depth=6,
                l2_leaf_reg=5.0,
                random_strength=.5,
            )

            if trial >= 4:
                p.update(
                    iterations=int(
                        rng.choice([700, 1100, 1600])
                    ),
                    learning_rate=float(
                        rng.choice([.025, .04, .06])
                    ),
                    depth=int(
                        rng.choice([5, 6, 7, 8])
                    ),
                    l2_leaf_reg=float(
                        rng.choice([3.0, 5.0, 10.0])
                    ),
                    random_strength=float(
                        rng.choice([0.0, .5, 1.0])
                    ),
                )

        name = (
            f"{family}_{trial:02d}_"
            f"{'log_L2' if log else 'raw_MAE'}_"
            f"{'rich' if rich else 'base'}"
        )

        configs[name] = dict(
            family=family,
            encoding=family,
            rich=rich,
            log=log,
            params=p,
        )

records = []

for name, spec in configs.items():
    print("\nCV:", name, flush=True)

    oof = np.full(len(yfit), np.nan)
    fold_mae = []
    t0 = time.perf_counter()

    for fold, (tr, va) in enumerate(cv, 1):
        model = PriceModel(spec).fit(
            Xfit.iloc[tr], yfit.iloc[tr]
        )

        oof[va] = model.predict(Xfit.iloc[va])

        score = mean_absolute_error(
            yfit.iloc[va], oof[va]
        )
        fold_mae.append(score)

        print(
            f"  Fold {fold}: MAE={score:.4f}",
            flush=True,
        )
        del model

    row = dict(
        model=name,
        family=spec["family"],
        **metrics(yfit, oof),
        fold_MAE_std=float(
            np.std(fold_mae, ddof=1)
        ),
        seconds=time.perf_counter() - t0,
    )
    records.append(row)

    (
        pd.DataFrame(records)
        .sort_values("MAE")
        .to_csv(OUT / "search_cv.csv", index=False)
    )

    gc.collect()

leaderboard = (
    pd.DataFrame(records)
    .sort_values("MAE")
    .reset_index(drop=True)
)

print(
    "\nDEVELOPMENT CV — candidate-selection scores, "
    "not independent test scores"
)
display(leaderboard.round(4))


# -------------------- 5. PREVIOUS VOTING RECIPE --------------------
# Refit from scratch, not the old already-fitted pickle.
# Otherwise the previous model would have seen the new
# blend-validation labels.

legacy = {
    "lgb": dict(
        family="lgb",
        encoding="ohe",
        rich=False,
        log=True,
        params=dict(
            n_estimators=350,
            learning_rate=.05,
            num_leaves=31,
            min_child_samples=30,
            subsample=.9,
            subsample_freq=1,
            colsample_bytree=.9,
            reg_lambda=1.0,
        ),
    ),
    "xgb": dict(
        family="xgb",
        encoding="ohe",
        rich=False,
        log=True,
        params=dict(
            n_estimators=350,
            learning_rate=.05,
            max_depth=6,
            min_child_weight=3,
            subsample=.9,
            colsample_bytree=.9,
        ),
    ),
    "cat": dict(
        family="cat",
        encoding="ohe",
        rich=False,
        log=True,
        params=dict(
            iterations=300,
            learning_rate=.08,
            depth=8,
            loss_function="RMSE",
            l2_leaf_reg=1.0,
            random_strength=0.0,
        ),
    ),
    "rf": dict(
        family="rf",
        encoding="ohe",
        rich=False,
        log=True,
        params=dict(
            n_estimators=200,
            max_depth=22,
            min_samples_leaf=2,
            max_features=.8,
        ),
    ),
}


def fit_legacy(Xtrain, ytrain):
    models = {
        n: PriceModel(s).fit(Xtrain, ytrain)
        for n, s in legacy.items()
    }
    return BlendModel(
        models,
        {n: 1 / len(models) for n in models},
    )


selected_names = []

for family in ["lgb", "cat"]:
    selected_names += (
        leaderboard.loc[
            leaderboard.family == family,
            "model",
        ]
        .head(TOP_MODELS_PER_FAMILY)
        .tolist()
    )

val_models = {}
val_predictions = {}

for name in selected_names:
    print(
        "Fitting blend-validation candidate:",
        name,
        flush=True,
    )

    val_models[name] = PriceModel(
        configs[name]
    ).fit(Xfit, yfit)

    val_predictions[name] = (
        val_models[name].predict(Xblend)
    )

print(
    "Fitting previous Voting recipe on "
    "the SAME training subset.",
    flush=True,
)

val_models["previous_voting"] = fit_legacy(
    Xfit, yfit
)
val_predictions["previous_voting"] = (
    val_models["previous_voting"].predict(Xblend)
)


# --------------------- 6. VALIDATION-ONLY BLENDING ----------------
names = list(val_predictions)

P = np.column_stack(
    [val_predictions[n] for n in names]
)
truth = yblend.to_numpy()

single_mae = np.mean(
    np.abs(P - truth[:, None]),
    axis=0,
)
best_single = int(np.argmin(single_mae))
single_weights = np.eye(len(names))[best_single]

best_weights = single_weights.copy()
best_loss = float(single_mae[best_single])

# Greedy convex ensemble:
# all weights >= 0 and sum to 1.
# No extrapolating stacker.
running = np.zeros(len(truth))
counts = np.zeros(len(names))

for step in range(1, BLEND_STEPS + 1):
    losses = np.mean(
        np.abs(
            (running[:, None] + P) / step
            - truth[:, None]
        ),
        axis=0,
    )

    j = int(np.argmin(losses))
    running += P[:, j]
    counts[j] += 1

    if losses[j] < best_loss:
        best_loss = float(losses[j])
        best_weights = counts.copy() / step

if (
    single_mae[best_single] - best_loss
    < MIN_BLEND_GAIN
):
    best_weights = single_weights

weights = {
    n: float(w)
    for n, w in zip(names, best_weights)
    if w > 0
}

validation_table = pd.DataFrame(
    [
        {
            "model": n,
            **metrics(yblend, val_predictions[n]),
        }
        for n in names
    ]
    + [
        {
            "model": "FROZEN_SELECTION",
            **metrics(yblend, P @ best_weights),
        }
    ]
)

print("\nBLEND VALIDATION — selection only")
display(validation_table.round(4))
print("Frozen weights:", weights)

validation_table.to_csv(
    OUT / "blend_validation_metrics.csv",
    index=False,
)

pd.DataFrame(
    {
        "csv_row_index": Xblend.index,
        "actual_price": truth,
        **val_predictions,
    }
).to_csv(
    OUT / "blend_validation_predictions.csv",
    index=False,
)

with open(OUT / "frozen_weights.json", "w") as f:
    json.dump(weights, f, indent=2)

del val_models
gc.collect()


# --------------------- 7. REFIT ON ALL DEVELOPMENT ----------------
# Includes blend-validation rows now that choices are frozen.
print(
    "\nRefitting previous Voting benchmark "
    "on ALL development rows.",
    flush=True,
)

previous_model = fit_legacy(Xd, yd)
final_parts = {}

for name in weights:
    print(
        "Refitting selected component:",
        name,
        flush=True,
    )

    final_parts[name] = (
        previous_model
        if name == "previous_voting"
        else PriceModel(configs[name]).fit(Xd, yd)
    )

final_model = BlendModel(final_parts, weights)


# --------------------- 8. SAME-HOLDOUT COMPARISON -----------------
# Do NOT select a different model after reading this table.
p_old = previous_model.predict(Xt)
p_new = final_model.predict(Xt)

results = pd.DataFrame(
    [
        {
            "model": "Previous Voting recipe — retrained",
            "rows": len(yt),
            **metrics(yt, p_old),
        },
        {
            "model": "V2 validation-selected model",
            "rows": len(yt),
            **metrics(yt, p_new),
        },
    ]
)

print(
    "\nFINAL COMPARISON: SAME CSV, "
    "SAME HOLDOUT, ALL VALID PRICES"
)
display(results.round(5))

for metric in ["MAE", "RMSE"]:
    old, new = results[metric].to_numpy()
    change = old - new

    status = (
        "unchanged"
        if np.isclose(old, new, rtol=1e-9, atol=1e-7)
        else ("improved" if change > 0 else "worse")
    )

    print(
        f"{metric}: {old:.5f} -> {new:.5f}; "
        f"reduction={change:.5f} ({status})"
    )

results.to_csv(
    OUT / "same_holdout_comparison.csv",
    index=False,
)

# Optional historical check:
# align prior predictions by row, ID and raw input.
# If the CSV changed, refuse to call it a verified
# historical comparison.
prior_predictions = (
    Path(PREVIOUS_DIR) / "holdout_predictions.csv"
    if PREVIOUS_DIR else None
)

if prior_predictions and prior_predictions.exists():
    old = pd.read_csv(
        prior_predictions,
        dtype={"id": "string"},
        low_memory=False,
    )
    old = old.set_index("csv_row_index")

    matched = (
        old.index.is_unique
        and set(old.index) == set(Xt.index)
    )

    if matched:
        old = old.loc[Xt.index]

        matched = np.allclose(
            old["actual_price"],
            yt,
            rtol=0,
            atol=1e-8,
        )

        if "id" in old:
            matched = (
                matched
                and old["id"].fillna("").equals(
                    ids.loc[Xt.index]
                    .astype("string")
                    .fillna("")
                    .rename("id")
                )
            )

        for c in NUM:
            if c in old:
                matched = (
                    matched
                    and np.allclose(
                        pd.to_numeric(
                            old[c], errors="coerce"
                        ),
                        pd.to_numeric(
                            Xt[c], errors="coerce"
                        ),
                        equal_nan=True,
                    )
                )

        for c in CAT:
            if c in old:
                matched = (
                    matched
                    and np.array_equal(
                        old[c]
                        .astype("string")
                        .fillna("")
                        .to_numpy(),
                        Xt[c]
                        .astype("string")
                        .fillna("")
                        .to_numpy(),
                    )
                )

    if matched:
        historical = pd.DataFrame(
            [
                {
                    "model": (
                        "Verified previous saved predictions"
                    ),
                    **metrics(
                        yt, old["predicted_price"]
                    ),
                },
                {
                    "model": "V2 frozen model",
                    **metrics(yt, p_new),
                },
            ]
        )

        print(
            "\nVerified historical holdout comparison:"
        )
        display(historical.round(5))

        historical.to_csv(
            OUT / "historical_holdout_comparison.csv",
            index=False,
        )

    else:
        print(
            "Historical file does not match this "
            "CSV/holdout. Historical comparison skipped."
        )


# --------------------- 9. SAVE EXACT EVALUATED MODEL --------------
versions = {
    p: metadata.version(p)
    for p in minimum
}

bundle = {
    "model": final_model,
    "weights": weights,
    "core_features": CORE,
    "selected_configs": {
        n: configs[n]
        for n in weights
        if n in configs
    },
    "legacy_configs": (
        legacy if "previous_voting" in weights else {}
    ),
    "versions": versions,
    "python_version": sys.version,
    "seed": SEED,
    "group_by_host": GROUP_BY_HOST,
}

with open(OUT / "model_bundle.pkl", "wb") as f:
    cloudpickle.dump(bundle, f)

with open(
    OUT / "previous_recipe_model.pkl", "wb"
) as f:
    cloudpickle.dump(previous_model, f)

digest = hashlib.sha256()

with open(DATA_PATH, "rb") as f:
    for chunk in iter(
        lambda: f.read(1024 * 1024),
        b"",
    ):
        digest.update(chunk)

file_hash = digest.hexdigest()

with open(OUT / "run_metadata.json", "w") as f:
    json.dump(
        {
            "csv_sha256": file_hash,
            "weights": weights,
            "versions": versions,
            "python_version": sys.version,
            "seed": SEED,
            "group_by_host": GROUP_BY_HOST,
            "holdout_rows": len(yt),
            "development_rows": len(yd),
            "all_search_configs": configs,
            "legacy_configs": legacy,
            "cv_folds": CV_FOLDS,
            "blend_validation_fraction": (
                BLEND_VALIDATION_SIZE
            ),
            "blend_steps": BLEND_STEPS,
            "min_blend_gain": MIN_BLEND_GAIN,
        },
        f,
        indent=2,
    )

(OUT / "requirements.txt").write_text(
    "\n".join(
        f"{k}=={v}" for k, v in versions.items()
    ) + "\n"
)


def predict_new(raw_data, saved=bundle):
    if isinstance(raw_data, dict):
        raw_data = pd.DataFrame([raw_data])
    elif not isinstance(raw_data, pd.DataFrame):
        raw_data = pd.DataFrame(raw_data)

    return pd.DataFrame(
        {
            "predicted_price": (
                saved["model"].predict(raw_data)
            )
        },
        index=raw_data.index,
    )


with open(OUT / "model_bundle.pkl", "rb") as f:
    # Load only trusted pickle files.
    reloaded = cloudpickle.load(f)

np.testing.assert_allclose(
    predict_new(Xt.head(5)),
    predict_new(Xt.head(5), reloaded),
)


# --------------------- 10. RESIDUALS AND PLOTS --------------------
diagnostics = Xt.copy()
diagnostics.insert(0, "id", ids.loc[Xt.index])

diagnostics["actual_price"] = yt
diagnostics["previous_prediction"] = p_old
diagnostics["predicted_price"] = p_new
diagnostics["absolute_error"] = abs(
    yt.to_numpy() - p_new
)
diagnostics["residual"] = yt.to_numpy() - p_new

diagnostics.to_csv(
    OUT / "holdout_predictions.csv",
    index_label="csv_row_index",
)

bands = pd.cut(
    yt,
    [0, 100, 200, 500, 1000, 2500, np.inf],
)

by_band = (
    diagnostics
    .groupby(bands, observed=True)
    .agg(
        rows=("absolute_error", "size"),
        MAE=("absolute_error", "mean"),
        mean_residual=("residual", "mean"),
    )
)

print(
    "\nDescriptive price-band errors; "
    "no rows removed:"
)
display(by_band.round(3))

by_band.to_csv(
    OUT / "price_band_errors.csv"
)

plt.figure(figsize=(7, 6))
plt.scatter(yt, p_new, alpha=.35, s=10)

limit = max(
    float(yt.max()),
    float(p_new.max()),
)

plt.plot(
    [0, limit],
    [0, limit],
    linestyle="--",
)

plt.xscale("symlog", linthresh=1)
plt.yscale("symlog", linthresh=1)
plt.xlabel("Actual price (symlog)")
plt.ylabel("Predicted price (symlog)")
plt.title(
    "V2: complete holdout, including expensive listings"
)
plt.tight_layout()
plt.savefig(
    OUT / "actual_vs_predicted.png",
    dpi=150,
)
plt.show()

print("\nSaved results:", OUT)
print(
    "Inference: predict_new(your_raw_dataframe)"
)
print(
    f"Actual elapsed time: "
    f"{time.perf_counter() - started:.1f} seconds"
)
print(
    "The old holdout has already been examined; "
    "this is a comparison, not a new external test."
)
print(
    "Lower MAE is not guaranteed, "
    "and lower MAE need not mean lower RMSE."
)


# Fresh-session inference, with recorded dependency versions:
#
# import cloudpickle
#
# with open(
#     '/content/airbnb_results_v2/model_bundle.pkl', 'rb'
# ) as f:
#     saved = cloudpickle.load(f)  # Trusted files only.
#
# prices = saved['model'].predict(new_raw_dataframe)
```
